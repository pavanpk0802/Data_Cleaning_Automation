```python
import pandas as pd
import numpy as np

# Create messy dataset
df = pd.DataFrame({
    "Name": ["Megha", "Rahul", "Anu", "Megha", "  Kiran  ", None],
    "Age": [21, 22, None, 21, 25, 30],
    "City": ["Bengaluru", "Mumbai", "Delhi", "Bengaluru", "  Chennai ", "Delhi"],
    "Salary": [30000, 45000, 50000, 30000, None, 60000]
})

print("Original Dataset:")
display(df)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Automated data cleaning

# Remove extra spaces from text columns
df["Name"] = df["Name"].str.strip()
df["City"] = df["City"].str.strip()

# Fill missing Name and City
df["Name"] = df["Name"].fillna("Unknown")
df["City"] = df["City"].fillna("Unknown")

# Fill missing Age with median age
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Salary with median salary
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

# Remove duplicate rows
df = df.drop_duplicates()

# Reset index
df = df.reset_index(drop=True)

print("Data cleaning completed successfully!")
display(df)
```

```python
# Validate cleaned dataset

print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nFinal Dataset:")
display(df)
```

```python
df.to_csv("cleaned_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
